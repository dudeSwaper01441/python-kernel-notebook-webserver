## 03 · Rich output

`display()` understands HTML, SVG, Markdown and images, and matplotlib figures render inline automatically when the library is installed.

This chart is generated by Python and rendered natively in the browser:

```python
from math import sqrt

values = [(n, n * n) for n in range(1, 8)]
peak = max(value for _, value in values)
bars = []

for i, (label, value) in enumerate(values):
    height = round(96 * sqrt(value / peak))
    x = 18 + i * 60
    bars.append(
        f'<rect x="{x}" y="{132 - height}" width="34" height="{height}" rx="7" fill="url(#bar)"/>'
        f'<text x="{x + 17}" y="{124 - height}" text-anchor="middle" fill="#c7d2fe" font-size="11">{value}</text>'
        f'<text x="{x + 17}" y="152" text-anchor="middle" fill="#8b98ab" font-size="10">{label}²</text>'
    )

display(HTML(f"""
<div style="padding:16px 8px 2px;font-family:system-ui,sans-serif">
  <div style="font-size:12.5px;color:#8b98ab;margin:0 0 8px 10px">n² — computed in Python, drawn in the browser</div>
  <svg viewBox="0 0 440 166" width="100%" style="max-width:440px;overflow:visible">
    <defs>
      <linearGradient id="bar" x1="0" y1="0" x2="0" y2="1">
        <stop offset="0%" stop-color="#9d8cff"/>
        <stop offset="100%" stop-color="#22d3ee"/>
      </linearGradient>
    </defs>
    <line x1="10" y1="132" x2="432" y2="132" stroke="rgba(148,163,184,.22)"/>
    {''.join(bars)}
  </svg>
</div>
"""))
```
