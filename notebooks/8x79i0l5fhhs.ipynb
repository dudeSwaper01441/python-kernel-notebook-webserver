**ตัวอย่างที่ 3:** ใช้ for loop เพื่อหาผลรวมของเลขจำนวนเต็มตั้งแต่ 1 ถึง n

````python
n = int(input())

# คำนวณผลรวมโดยใช้ for loop
total = 0
for i in range(1, n + 1):
    total += i

# แสดงผลลัพธ์
print(total)
````

# **การใช้คำสั่ง break และ continue ใน loop**

**การใช้ break**
- คำสั่ง break ใช้เพื่อหยุดการทำงานของ loop ทั้งหมด โดยทันทีที่เจอ break โปรแกรมจะออกจาก loop ทันที

**การใช้ continue**
- คำสั่ง continue ใช้เพื่อข้ามไปยังรอบถัดไปของ loop โดยจะไม่ทำคำสั่งที่เหลือในรอบนั้น ๆ

**ตัวอย่างที่ 7:** หยุดการวนลูปเมื่อเจอเลข 5

````python
for i in range(1, 11):
    if i == 5:
        break  # หยุด loop เมื่อ i เท่ากับ 5
    print(i)
````

**ตัวอย่างที่ 8:** ข้ามการแสดงเลขคู่

````python
for i in range(1, 11):
    if i % 2 == 0:
        continue  # ข้ามเมื่อ i เป็นเลขคู่
    print(i)
````
